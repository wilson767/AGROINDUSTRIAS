```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
X = np.array([
    [1, 5, 1],  # Floral, Eau de Parfum, Femenino
    [2, 3, 0],  # Cítrico, Eau de Toilette, Unisex
    [3, 4, 1],  # Amaderado, Eau de Parfum, Femenino
    [4, 2, 2],  # Oriental, Eau de Cologne, Masculino
    [2, 3, 1],  # Cítrico, Eau de Toilette, Femenino
    [3, 2, 2],  # Amaderado, Eau de Cologne, Masculino
    [4, 5, 0],  # Oriental, Eau de Parfum, Unisex
    [1, 4, 1],  # Floral, Eau de Parfum, Femenino
    [2, 3, 2],  # Cítrico, Eau de Toilette, Masculino
    [3, 1, 2]   # Amaderado, Eau Fraîche, Masculino
])

y = np.array(["Dior", "Chanel", "Gucci", "Dolce&Gabbana", "Versace", "Tom Ford", "YSL", "Givenchy", "Hermès", "Armani"])
```

```python
df = pd.DataFrame(X, columns=["Nota Olfativa", "Concentración", "Género"])
df["Perfume"] = y
```

```python
sns.scatterplot(x="Nota Olfativa", y="Concentración", hue="Perfume", data=df)
plt.title("Distribución de Perfumes por Notas y Concentración")
plt.show()
```

```python
# Paso 5: Crear y entrenar el modelo
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Paso 6: Hacer predicciones
modelo_knn = KNeighborsClassifier(n_neighbors=3)
modelo_knn.fit(X_train, y_train)
```

```python
# Paso 7: Evaluar el modelo
y_pred = modelo_knn.predict(X_test)
```

```python
# Paso 8: Probar con un nuevo dato
print("Precisión:", accuracy_score(y_test, y_pred))
print("\nReporte de Clasificación:\n", classification_report(y_test, y_pred))
```

```python
conf_matrix = confusion_matrix(y_test, y_pred)
sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues")
plt.xlabel("Predicción")
plt.ylabel("Real")
plt.title("Matriz de Confusión")
plt.show()
```

```python
nuevo_perfume = np.array([[2, 4, 1]])  # Perfume cítrico, Eau de Parfum, Femenino
prediccion = modelo_knn.predict(nuevo_perfume)
print("Perfume recomendado:", prediccion[0])
```
